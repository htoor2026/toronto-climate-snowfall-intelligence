```python
from pathlib import Path
import pandas as pd
import duckdb

CSV_PATH = Path("../data/raw/ahccd/ahccd.csv")

df = pd.read_csv(CSV_PATH)
df["time"] = pd.to_datetime(df["time"])

con = duckdb.connect("../database/climate.duckdb")
```

```python
con.register("df_view", df)
```

```python
con.execute("""
CREATE OR REPLACE TABLE climate_daily AS
SELECT * FROM df_view
""")
```

```python
con.execute("SHOW TABLES").fetchdf()
```

```python
con.execute("""
SELECT *
FROM climate_daily
LIMIT 5
""").fetchdf()
```
